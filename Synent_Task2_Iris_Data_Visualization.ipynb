# Task 2: Data Visualization Using the Iris Dataset

## Objective
The objective of this project is to visualize patterns and relationships in the Iris dataset using different types of charts.

Data visualization helps us understand distributions, compare categories, and identify relationships between variables. In this project, I will create a bar chart, histogram, and scatter plot to explore the characteristics of three Iris flower species.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Pandas is used to organize and inspect the dataset. Matplotlib and Seaborn are used to create charts and visualize patterns. These libraries help present data in a clear and understandable format.

```python
df = sns.load_dataset("iris")

df.head()
```

The Iris dataset is loaded from Seaborn into a Pandas DataFrame. It contains measurements of sepal length, sepal width, petal length, and petal width for three species: setosa, versicolor, and virginica. The `head()` function displays the first five records.

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nSpecies Count:")
print(df["species"].value_counts())
```

This step checks the dataset's dimensions, column names, and missing values. The species count shows how many flower samples belong to each species. These checks help us understand the data before creating visualizations.

```python
species_counts = df["species"].value_counts()

species_counts = species_counts.reindex(
    ["setosa", "versicolor", "virginica"]
)

species_counts.plot(kind="bar", rot=0)

plt.title("Number of Samples by Iris Species")
plt.xlabel("Iris Species")
plt.ylabel("Number of Samples")
plt.tight_layout()
plt.show()
```

A bar chart is used to compare the number of samples belonging to each Iris species. The species are displayed on the horizontal axis, while the number of samples is displayed on the vertical axis. This visualization helps us understand the distribution of samples across categories.

```python
plt.figure(figsize=(8, 5))

plt.hist(df["sepal_length"], bins=10, edgecolor="black")

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

A histogram shows the distribution of sepal length measurements by grouping values into intervals called bins. The horizontal axis represents sepal length in centimetres, and the vertical axis represents the frequency of observations in each interval. This helps identify common measurement ranges and the overall distribution.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species",
    s=70
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.tight_layout()
plt.show()
```

A scatter plot displays the relationship between petal length and petal width. Each point represents one flower, and different colours distinguish the three species. This visualization helps identify relationships between measurements and determine whether the species form distinct groups.

```python
feature_comparison = df.groupby("species")[
    ["sepal_length", "sepal_width",
     "petal_length", "petal_width"]
].mean().round(2)

feature_comparison
```

This step calculates the average sepal length, sepal width, petal length, and petal width for each species. Comparing these averages helps identify differences in flower measurements and provides numerical support for the visual observations.

```python
df.to_csv("iris_visualization_dataset.csv", index=False)

print("Dataset exported successfully!")
```

The Iris dataset is exported to a CSV file using Pandas.

## Conclusion

The Iris dataset was explored using a bar chart, histogram, and scatter plot. The bar chart compared sample counts across species, the histogram showed the distribution of sepal length, and the scatter plot illustrated the relationship between petal length and petal width. Comparing average measurements also helped identify differences among the three species.
